### 総費用の追加→不採用

```python
df = pd.read_csv("/home/haruk/workspace/kaggle-ishiwata/kaggle/new_compettition/data/Hotel Reservations.csv")
X_train,X_test,y_train,y_test = preprocessing_pipeline(
    df,new_columns,columns_mappings_dict,categorical_cols
)

X_train["合計宿泊数"] = X_train["平日の宿泊数"] + X_train["週末の宿泊数"]
X_test["合計宿泊数"] = X_test["平日の宿泊数"] + X_test["週末の宿泊数"]

X_train["宿泊人数"] = X_train["大人の数"] + X_train["子供の数"]
X_test["宿泊人数"] = X_test["大人の数"] + X_test["子供の数"]

X_train["総費用"] = X_train["一泊あたりの宿泊費"] * X_train["合計宿泊数"]
X_test["総費用"] = X_test["一泊あたりの宿泊費"] * X_test["合計宿泊数"]
```

```python
lr_model, lr_report, lr_coefficients = fit_predict_evaluate(X_train,X_test,y_train,y_test)

# 評価レポートの表示
print("\n=== ロジスティック回帰 評価レポート ===")
print(lr_report)

# 特徴量の影響度（係数）トップ10の表示
print("\n=== 特徴量の影響度（係数）トップ10 ===")
lr_coefficients
```
