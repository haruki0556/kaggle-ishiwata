```python
import pandas as pd
import numpy as np

df = pd.read_csv("/home/haruk/workspace/kaggle-ishiwata/kaggle/new_compettition/data/Hotel Reservations.csv")

new_columns = {
"Booking_ID":"予約ID",
"no_of_adults":"大人の数",
"no_of_children":"子供の数",
"no_of_weekend_nights":"週末の宿泊数",
"no_of_week_nights":"平日の宿泊数",
"type_of_meal_plan":"食事プランの種類",
"required_car_parking_space":"駐車場の要不要",
"room_type_reserved":"予約された部屋のタイプ",
"lead_time":"予約日から到着日までの日数",
"arrival_year":"到着年",
"arrival_month":"到着月",
"arrival_date":"到着日",
"market_segment_type":"マーケットセグメント",
"repeated_guest":"リピーター客",
"no_of_previous_cancellations":"過去のキャンセル回数",
"no_of_previous_bookings_not_canceled":"過去のキャンセルされなかった予約数",
"avg_price_per_room":"一泊あたりの宿泊費",
"no_of_special_requests":"特別リクエストの数",
"booking_status":"予約ステータス",
}

df_renamed = df.rename(columns=new_columns)

#予約IDの削除
df_renamed.drop("予約ID",inplace=True,axis=1)

#食事プランの種類の中身を日本語に変換
meal_plan_mapping = {
    'Meal Plan 1': 'プラン1',
    'Meal Plan 2': 'プラン2',
    'Meal Plan 3': 'プラン3',
    'Not Selected': '選択なし'
}

df_renamed['食事プランの種類'] = df_renamed['食事プランの種類'].replace(meal_plan_mapping)

#部屋タイプを日本後に変換
room_type_mapping = {
'Room_Type 1':"タイプ1", 
'Room_Type 4':'タイプ4',
'Room_Type 2':"タイプ2",
'Room_Type 6':"タイプ6",
'Room_Type 5':"タイプ5",
'Room_Type 7':"タイプ7", 
'Room_Type 3':"タイプ3",
}

df_renamed["予約された部屋のタイプ"] = df_renamed["予約された部屋のタイプ"].replace(room_type_mapping)

#マーケットセグメントの日本語変換
segment_mapping = {
    "Offline":"オフライン予約",
    "Online":"オンライン予約",
    "Corporate":"法人予約",
    "Aviation":"航空",
    "Complementary":"招待",
}

df_renamed["マーケットセグメント"] = df_renamed["マーケットセグメント"].replace(segment_mapping)

#目的変数のクラスラベル化
status_mapping = {
    "Not_Canceled":0,
    "Canceled":1,
}

df_renamed["予約ステータス"] = df_renamed["予約ステータス"].replace(status_mapping)

df_renamed.head()
```
