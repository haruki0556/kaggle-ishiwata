カラム名 (英語),日本語訳,推測されるデータの意味
Hours_Studied,学習時間,勉強に費やした時間。
Attendance,出席率/出席状況,授業への参加度合い。
Parental_Involvement,親の関与度,親が子どもの学習にどれだけ関わっているか。
Access_to_Resources,学習資源へのアクセス,教材、図書館、インターネットなどの利用しやすさ。
Extracurricular_Activities,課外活動,スポーツやクラブ活動への参加状況。
Sleep_Hours,睡眠時間,毎日の睡眠時間。
Previous_Scores,以前の成績,過去のテストや評価の点数。
Motivation_Level,モチベーションレベル,学習意欲の度合い。
Internet_Access,インターネットアクセス,自宅などでのインターネット接続環境の有無。
Tutoring_Sessions,個別指導セッション,家庭教師や補習などを受けた回数/有無。
Family_Income,世帯収入,家族の経済状況。
Teacher_Quality,教師の質,教師の評価や指導能力。
School_Type,学校の種類,公立、私立、男女共学、別学など。
Peer_Influence,仲間からの影響,友人や同級生からの学習意欲や習慣への影響。
Physical_Activity,身体活動,運動の頻度や量。
Learning_Disabilities,学習障害,ディスレクシアなどの学習に関する障害の有無。
Parental_Education_Level,親の学歴レベル,両親の最終学歴。
Distance_from_Home,自宅からの距離,学校までの距離（通学時間など）。
Gender,性別,生徒の性別。
Exam_Score,試験の点数,**目的変数（ターゲット）**となる試験の最終成績。
